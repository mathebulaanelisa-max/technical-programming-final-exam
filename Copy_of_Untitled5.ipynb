```python
party_votes = df.groupby('PartyName')['TotalValidVotes'].sum()
print(party_votes.sort_values(ascending=False))
```

```python
party_share = (party_votes / party_votes.sum()) * 100
print(party_share.sort_values(ascending=False))
```

```python
ward_party_votes = df.groupby(
    ['Ward', 'PartyName']
)['TotalValidVotes'].sum()

print(ward_party_votes)
```

```python
ward_votes = df.groupby('Ward')['TotalValidVotes'].sum()
print(ward_votes)
```

```python
ward_party_table = df.groupby(
    ['Ward', 'PartyName']
)['TotalValidVotes'].sum().reset_index()

ward_leaders = ward_party_table.loc[
    ward_party_table.groupby('Ward')['TotalValidVotes'].idxmax()
]

print(ward_leaders)
```

```python
ward_party_table['WardVoteShare'] = (
    ward_party_table['TotalValidVotes'] /
    ward_party_table.groupby('Ward')['TotalValidVotes'].transform('sum')
) * 100
```

```python
print(ward_party_table)
```

```python
second_share = ward_party_table.groupby('Ward')['WardVoteShare'].apply(
    lambda x: x.sort_values(ascending=False).iloc[1]
    if len(x) > 1 else 0
)
print(second_share)
```

```python
highest_share = ward_party_table.groupby('Ward')['WardVoteShare'].max()
print(highest_share)
```

```python
ward_competition = highest_share - second_share

print(ward_competition.sort_values())
```
